```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. CARREGAMENTO DO DATASET PELO GITHUB

```python
url = (
    "https://raw.githubusercontent.com/"
    "rannats/Intro-IA/main/Atividade-01/"
    "qualidade_ar_real_uci.csv"
)

df = pd.read_csv(
    url,
    parse_dates=["data_hora"]
)

print("Dataset carregado com sucesso!")
print(f"Número de linhas: {df.shape[0]}")
print(f"Número de colunas: {df.shape[1]}")

```

## 6. GRÁFICO DA CONCENTRAÇÃO DE CO AO LONGO DO TEMPO

```python
plt.figure(figsize=(14, 5))

plt.plot(
    df["data_hora"],
    df["CO(GT)"],
    color="blue",
    linewidth=0.7,
    label="Concentração de CO"
)

plt.title("Concentração de monóxido de carbono ao longo do tempo")
plt.xlabel("Data")
plt.ylabel("CO (mg/m³)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## Observações técnicas

- A concentração de monóxido de carbono apresentou variações ao longo do período analisado, incluindo alguns picos bem superiores aos valores mais frequentes.

- A média da concentração de CO foi de aproximadamente 2,15 mg/m³, enquanto o valor máximo registrado chegou a 11,9 mg/m³.

- O dataset apresenta valores ausentes na variável CO(GT), o que indica a necessidade de realizar um tratamento dos dados antes de utilizá-los em modelos de inteligência artificial.
